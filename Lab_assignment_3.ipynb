```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Data Generation

```python

np.random.seed(42)
regular_wait = np.random.uniform(2, 12, 3000)

busy_wait = np.random.normal(8, 2, 3000)
busy_wait = np.where(busy_wait < 0, 0, busy_wait)
```

# Visualization

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(regular_wait, bins=30, kde=True)
plt.title("Regular Day Wait Times (Uniform)")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
sns.histplot(busy_wait, bins=30, kde=True)
plt.title("Busy Day Wait Times (Normal)")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

# Statistical Analysis

```python
print("Regular Day Statistics:")
regular_mean = np.mean(regular_wait)
regular_median = np.median(regular_wait)
regular_std = np.std(regular_wait)
regular_max = np.max(regular_wait)

print("Mean:", regular_mean)
print("Median:", regular_median)
print("Standard Deviation:", regular_std)
print("Maximum:", regular_max)


print("\nBusy Day Statistics:")
busy_mean = np.mean(busy_wait)
busy_median = np.median(busy_wait)
busy_std = np.std(busy_wait)
busy_max = np.max(busy_wait)

print("Mean:", busy_mean)
print("Median:", busy_median)
print("Standard Deviation:", busy_std)
print("Maximum:", busy_max)
```
