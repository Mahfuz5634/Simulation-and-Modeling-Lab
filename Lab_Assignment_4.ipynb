```python
import random
import math

def estimate_pi(n):
    inside = 0
    for _ in range(n):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x*x + y*y <= 1:
            inside += 1
    pi_val = 4 * inside / n
    return inside, pi_val

sizes = [100, 1000, 10000, 100000]

print(f"{'Total':<10}{'Inside':<10}{'Estimated_pi':<15}{'Error'}")

for n in sizes:
    inside, pi_val = estimate_pi(n)
    err = abs(math.pi - pi_val)
    print(f"{n:<10}{inside:<10}{pi_val:<15.6f}{err:.6f}")
```
